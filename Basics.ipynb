```python
import numpy as np
import pandas as pd
```

**Matplotlib Basics**

```python
import matplotlib.pyplot as plt
```

```python
x=np.linspace(0,5,11)
x
```

```python
y=x**3
y
```

```python
plt.plot(x,y)
plt.title("Simple Plot")
plt.xlabel("x")
plt.ylabel("y")
```

```python
plt.scatter(x,y)
plt.title("Basic Plot")
plt.xlabel("X_axis")
plt.ylabel("Y_axis")
```

**Subplots in Matplotlib**

```python
z=y*3
z
```

```python
plt.title("Intro to subplots")
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(z,z)
plt.subplot(2,2,3)
plt.plot(x,z)
plt.subplot(2,2,4)
plt.plot(y,z)
```

**Object Oreinted Way**

```python
fig=plt.figure(figsize=(20,8),dpi=100)
axis1=fig.add_axes([0.1,0.1,.4,.8])
axis1.plot(x,y)
axis2=fig.add_axes([0.3,0.3,0.1,0.1])
axis2.plot(x,y**2)
```
