**Categorical Data Plots**

Now let's discuss using seaborn to plot categorical data! There are a few main plot types for this:

boxplot
violinplot
stripplot
swarmplot
barplot
countplot

```python
sns.countplot(y=df['smoker'],hue=df['sex'])
```

```python
df['time'].unique()
```

```python
sns.countplot(x=df['time'],hue=df['sex'])
```

```python
df
```

```python
sns.countplot(data=df,x='time',hue='sex')
```

```python
sns.barplot(data=df,x='sex',y='total_bill')
```

```python
sns.boxplot(df['total_bill'])
```

```python
sns.barplot(data=df,x='total_bill',y='sex')
```

```python
sns.violinplot(data=df,y='sex',x='total_bill')
```

```python
sns.boxplot(df['total_bill'])
```
